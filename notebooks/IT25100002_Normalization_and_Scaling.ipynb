```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
import warnings
warnings.filterwarnings('ignore')

rfm = pd.read_csv('../results/outputs/step4_rfm_encoded.csv')

print("--- Technique: Normalization & Scaling ---")
# 1. Log Transformation to handle remaining right-skewness safely
rfm['Log_R'] = np.log1p(rfm['Recency'])
rfm['Log_F'] = np.log1p(rfm['Frequency'])
rfm['Log_M'] = np.log1p(rfm['Monetary'])

# 2. StandardScaler to bring mean to 0 and variance to 1
scaler = StandardScaler()
features_to_scale = ['Log_R', 'Log_F', 'Log_M']
rfm[['Scaled_R', 'Scaled_F', 'Scaled_M']] = scaler.fit_transform(rfm[features_to_scale])

print(rfm[['Scaled_R', 'Scaled_F', 'Scaled_M']].head())

print("\n--- EDA: Scaled Feature Distributions ---")
plt.figure(figsize=(12,4))
plt.subplot(1,3,1); sns.histplot(rfm['Scaled_R'], kde=True, color='blue'); plt.title("Scaled Recency")
plt.subplot(1,3,2); sns.histplot(rfm['Scaled_F'], kde=True, color='green'); plt.title("Scaled Frequency")
plt.subplot(1,3,3); sns.histplot(rfm['Scaled_M'], kde=True, color='red'); plt.title("Scaled Monetary")
plt.tight_layout()
plt.show()

rfm.to_csv('../results/outputs/step5_rfm_scaled.csv', index=False)
```
