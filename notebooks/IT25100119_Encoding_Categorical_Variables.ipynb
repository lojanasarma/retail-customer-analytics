```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

df = pd.read_csv('../results/outputs/step2_full_data_with_cost.csv')
rfm = pd.read_csv('../results/outputs/step3_rfm_no_outliers.csv')

print("--- Technique: Encoding Categorical Variables ---")
# We will One-Hot Encode the top 3 purchasing countries for our ML model
top_countries = ['United Kingdom', 'Germany', 'France']
# Filter data to top countries and get unique customers
df_filtered = df[df['Country'].isin(top_countries)][['Customer ID', 'Country']].drop_duplicates()

# One-Hot Encoding
encoded_countries = pd.get_dummies(df_filtered, columns=['Country'], prefix='Is')

# Merge encoded categorical features back with our numerical RFM data
rfm_encoded = pd.merge(rfm, encoded_countries, left_on='CustomerID', right_on='Customer ID', how='inner')
rfm_encoded.drop('Customer ID', axis=1, inplace=True)
print(rfm_encoded.head())

print("\n--- EDA: Categorical Correlation ---")
plt.figure(figsize=(8,6))
sns.heatmap(rfm_encoded.corr(), annot=True, cmap='coolwarm', fmt=".2f")
plt.title("Correlation including Encoded Categorical Variables")
plt.show()

rfm_encoded.to_csv('../results/outputs/step4_rfm_encoded.csv', index=False)
```
